```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px

# Display all columns
pd.set_option("display.max_columns", None)

# Load cleaned data
df = pd.read_csv("../data/clean_sales.csv")

# Convert dates again (CSV saves them as strings)
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

df.head()
```

```python
print("Rows :", df.shape[0])
print("Columns :", df.shape[1])

df.info()
```

```python
df.describe(include="all")
```

```python
total_sales = df["Sales"].sum()

average_sales = df["Sales"].mean()

total_orders = df["Order ID"].nunique()

total_customers = df["Customer ID"].nunique()

total_products = df["Product Name"].nunique()

total_states = df["State"].nunique()

total_cities = df["City"].nunique()

print(f"Total Sales      : {total_sales:,.2f}")
print(f"Average Sales    : {average_sales:,.2f}")
print(f"Orders           : {total_orders}")
print(f"Customers        : {total_customers}")
print(f"Products         : {total_products}")
print(f"States           : {total_states}")
print(f"Cities           : {total_cities}")
```

```python
monthly_sales = (
    df.groupby(["Year", "Month Name"], sort=False)["Sales"]
      .sum()
      .reset_index()
)

monthly_sales
```

```python
fig = px.line(
    monthly_sales,
    x="Month Name",
    y="Sales",
    color="Year",
    markers=True,
    title="Monthly Sales Trend"
)

fig.show()
```

```python
category_sales = (
    df.groupby("Category")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)

category_sales
```

```python
fig = px.bar(
    category_sales,
    x="Category",
    y="Sales",
    color="Category",
    text_auto=".2s",
    title="Sales by Category"
)

fig.show()
```

```python
region_sales = (
    df.groupby("Region")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)

region_sales
```

```python
fig = px.bar(
    region_sales,
    x="Region",
    y="Sales",
    color="Region",
    text_auto=".2s",
    title="Regional Sales"
)

fig.show()
```

```python
top_states = (
    df.groupby("State")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
      .reset_index()
)

top_states
```

```python
fig = px.bar(
    top_states,
    x="Sales",
    y="State",
    orientation="h",
    text_auto=".2s",
    title="Top 10 States by Sales"
)

fig.show()
```

```python
top_customers = (
    df.groupby("Customer Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
      .reset_index()
)

top_customers
```

```python
top_products = (
    df.groupby("Product Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
      .reset_index()
)

top_products
```

```python
segment_sales = (
    df.groupby("Segment")["Sales"]
      .sum()
      .reset_index()
)

fig = px.pie(
    segment_sales,
    names="Segment",
    values="Sales",
    title="Sales by Customer Segment",
    hole=0.5
)

fig.show()
```

```python
ship_mode = (
    df.groupby("Ship Mode")
      .agg(
          Orders=("Order ID", "count"),
          Sales=("Sales", "sum"),
          AvgShippingDays=("Shipping Days", "mean")
      )
      .reset_index()
)

ship_mode
```

```python
print("Highest Sales Category :", category_sales.iloc[0]["Category"])

print("Best Region :", region_sales.iloc[0]["Region"])

print("Top Customer :", top_customers.iloc[0]["Customer Name"])

print("Top Product :", top_products.iloc[0]["Product Name"])

print("Highest Sales State :", top_states.iloc[0]["State"])
```
